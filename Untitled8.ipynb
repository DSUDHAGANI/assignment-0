```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
# union — the two halves of the season share the same columns
sales = pd.concat([pd.read_csv("market_sales_part1.csv"),
pd.read_csv("market_sales_part2.csv")], ignore_index=True)
sales["date"] = pd.to_datetime(sales["date"])
# join — bring in city, county, latitude, longitude
markets = pd.read_csv("market_directory.csv")
df = sales.merge(markets, on="market_id", how="left")
print(df.shape)
```

```python
rev = df.groupby("category")["revenue"].sum().sort_values()

fig, ax = plt.subplots(figsize=(8, 3.6))
ax.barh(rev.index, rev.values / 1000, color="#1F4E8C")

for i, v in enumerate(rev.values / 1000):
    ax.text(v + 8, i, f"${v:,.0f}k", va="center", fontsize=9)

ax.set_xlim(0, rev.max() / 1000 * 1.15)
ax.set_xlabel("Total revenue, 2026 season (thousands of dollars)")
ax.set_title("Prepared Food out-earns every other category", fontweight="bold")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.savefig("ex41.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.dates as mdates

weekly = df.set_index("date")["revenue"].resample("W").sum()

peak_date = weekly.idxmax()
peak_value = weekly.max()
print("Peak week ending:", peak_date.date())
print("Revenue that week:", round(peak_value))

fig, ax = plt.subplots(figsize=(8.4, 3.5))
ax.plot(weekly.index, weekly.values / 1000, color="#2E8B8B", lw=2)

ax.scatter([peak_date], [peak_value / 1000], color="#D9472B", zorder=3)
ax.annotate(
    f"peak week\n{peak_date:%d %b} \u00b7 ${peak_value/1000:,.0f}k",
    xy=(peak_date, peak_value / 1000),
    xytext=(peak_date, peak_value / 1000 - 28),
    ha="center", color="#D9472B", fontsize=9,
    arrowprops=dict(arrowstyle="-", color="#D9472B"),
)

ax.set_ylim(0, 120)
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
ax.set_ylabel("Revenue per week ($000)")
ax.set_title("Season revenue builds to a mid-August peak", fontweight="bold")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.savefig("ex42.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
rated = df[df["customer_rating"] > 0]

order = (rated.groupby("category")["customer_rating"]
              .median().sort_values(ascending=False).index)
counts = rated.groupby("category").size()
means = rated.groupby("category")["customer_rating"].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8.2, 3.9))
sns.boxplot(data=rated, x="customer_rating", y="category", order=order,
            ax=ax, color="#C9D6E8", fliersize=3)

ax.set_xlim(0.8, 5.9)
for i, cat in enumerate(order):
    ax.text(5.25, i, f"n = {counts[cat]:,}", va="center", fontsize=8)

ax.set_xlabel("Customer rating (unrated rows excluded)")
ax.set_ylabel("")
ax.set_title(f"{means.index[0]} and {means.index[1]} rate highest; {means.index[-1]} trails",
             fontweight="bold")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

print("Mean including the zeros:", round(df["customer_rating"].mean(), 3))
print("Mean excluding them:", round(rated["customer_rating"].mean(), 3))
print("Rows removed:", len(df) - len(rated), "| Rows kept:", len(rated))

plt.savefig("ex43.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.dates as mdates

df["month"] = df["date"].dt.to_period("M").dt.to_timestamp()
panel = df.groupby(["category", "month"])["revenue"].sum().reset_index()

ref = panel.groupby("month")["revenue"].mean()
order = panel.groupby("category")["revenue"].sum().sort_values(ascending=False).index

fig, axes = plt.subplots(2, 3, figsize=(9.6, 4.9), sharex=True, sharey=True)
for k, (ax, cat) in enumerate(zip(axes.ravel(), order)):
    sub = panel[panel["category"] == cat]
    ax.plot(ref.index, ref.values / 1000, color="grey", ls="--", lw=1)
    ax.plot(sub["month"], sub["revenue"] / 1000, color="#1F4E8C", lw=2)
    ax.set_title(cat, fontsize=9, fontweight="bold", loc="left")
    ax.set_ylim(bottom=0)
    ax.xaxis.set_major_locator(mdates.MonthLocator(bymonth=[5, 7, 9]))
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%b"))
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    if k % 3 == 0:
        ax.set_ylabel("Revenue ($000)")

fig.suptitle("Every category peaks in August \u2014 but Prepared Food peaks hardest",
             fontweight="bold", fontsize=11)
fig.text(0.5, 0.93,
         "dashed grey line = average month across all six categories; all panels share one y-axis",
         ha="center", fontsize=8, color="grey")

fig.tight_layout(rect=[0, 0, 1, 0.92])
plt.savefig("ex44.png", dpi=150, bbox_inches="tight")
plt.show()
```
